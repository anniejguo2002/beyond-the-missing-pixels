# Beyond the Missing Pixels — results analysis

This notebook plots **previously reported summary measurements** shipped with this repository. It is **not the original training/evaluation Colab notebook**. Export that original notebook from Colab as `experiment.ipynb` and add it separately.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
root = Path.cwd()
results = root / "results"
final = pd.read_csv(results / "final_test_summary.csv")
mask = pd.read_csv(results / "mask_size_summary.csv")
display(final)
```

```python
ax = final.plot.bar(x="Method", y="Mean masked MAE", legend=False, figsize=(8,4))
ax.set_ylabel("Mean masked MAE (lower is better)")
ax.set_title("Held-out test: 3,669 images")
plt.tight_layout()
plt.show()
```

```python
ax=mask.plot(x="mask_size", y=["U-Net","Telea","Navier-Stokes"], marker="o", figsize=(8,4))
ax.set_xlabel("Square mask width (pixels)")
ax.set_ylabel("Masked MAE (lower is better)")
ax.set_title("Mask-size sensitivity: 100 validation images")
plt.tight_layout()
plt.show()
```
